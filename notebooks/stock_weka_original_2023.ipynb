```python
import weka.core.jvm as jvm
jvm.start()
from weka.classifiers import Classifier
```

```python
from weka.classifiers import Classifier
from weka.core.converters import Loader
from weka.core.dataset import Instance, Instances

# Load the historical stock price data
loader = Loader(classname="weka.core.converters.ArffLoader")
data = loader.load_file("/content/sample_data/INFY.csv")

# Create a linear regression model
model = Classifier(classname="weka.classifiers.functions.LinearRegression")
model.build_classifier(data)

# Create an instance with the values you want to classify
values = [2022, 15.0, 20.0, 10.0, 10000.0]
instance = Instance.create_instance(values)
instance.dataset = Instances.template_instances(data)

# Use the model to predict the closing price on a given date
prediction = model.classify_instance(instance)
print("Predicted closing price:", prediction)
```

```python
from weka.classifiers import Classifier
from weka.core.converters import Loader

# Load the historical stock price data
loader = Loader(classname="weka.core.converters.ArffLoader")
data = loader.load_file("/content/stock_prices.arff")

# Set the class attribute
class_index = data.num_attributes - 1
data.class_index = class_index




# Create a linear regression model
model = Classifier(classname="weka.classifiers.functions.LinearRegression")
model.build_classifier(data)

# Create an instance with the values you want to classify
values = [2022, 15.0, 20.0, 10.0, 10000.0]
instance = Instance.create_instance(values)
instance.dataset = Instances.template_instances(data)

# Use the model to predict the closing price on a given date
prediction = model.classify_instance(instance)
print("Predicted closing price:", prediction)
```
